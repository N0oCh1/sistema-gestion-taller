## Ventas diarias LogiPanamá S.A. — agregación diaria, semanal y mensual

```python
%pip install matplotlib
```

```python
import matplotlib.pyplot as plt

ventas = pd.read_csv('ventas_diarias_logipanama_capitulo2.csv', parse_dates=['Fecha'])
ventas = ventas.set_index('Fecha')

semanal = ventas.resample('W').sum()
# 'M' (fin de mes) está obsoleto desde pandas 2.2 y fue eliminado en versiones nuevas; 'ME' es el equivalente
mensual = ventas.resample('ME').sum()

print(ventas.head())
print(semanal.head())
print(mensual)
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 10), sharex=True)

axes[0].plot(ventas.index, ventas['Envios_Gestionados'], color='tab:blue', linewidth=0.8)
axes[0].set_title('Envíos diarios')

axes[1].plot(semanal.index, semanal['Envios_Gestionados'], color='tab:orange', marker='o', markersize=3)
axes[1].set_title('Envíos semanales (suma)')

axes[2].bar(mensual.index, mensual['Envios_Gestionados'], width=20, color='tab:green')
axes[2].set_title('Envíos mensuales (suma)')

for ax in axes:
    ax.set_ylabel('Envíos')
    ax.grid(alpha=0.3)
axes[2].set_xlabel('Fecha')

fig.suptitle('LogiPanamá S.A. — Envíos gestionados en 2025 por nivel de agregación', fontsize=14)
plt.tight_layout()
plt.show()
```

### Análisis de los patrones por nivel jerárquico

**Diario:** la serie muestra un fuerte "ruido" de un día a otro, pero es el único nivel donde se ve el **ciclo semanal**: los envíos suben de lunes a miércoles (≈1,460 envíos promedio el miércoles) y bajan en viernes, sábado y domingo (≈1,140 el sábado). También permite detectar días atípicos, como el pico del 24 de diciembre (2,058 envíos). Sin embargo, la tendencia de fondo cuesta verla entre tantas oscilaciones.

**Semanal:** al sumar por semana se elimina el ciclo de días laborables/fin de semana y la curva se suaviza, de modo que se distinguen mejor los **cambios de mediano plazo**: la caída gradual hacia mitad de año y la subida sostenida desde septiembre, con las semanas más altas en diciembre (más de 12,000 envíos). Hay que tener cuidado con la primera y la última semana, que están incompletas (5 y 3 días), por eso aparecen como caídas artificiales.

**Mensual:** es el nivel más agregado y muestra con claridad la **estacionalidad anual**: un primer trimestre estable (≈38,000–41,000 envíos), un valle en junio–julio (≈32,000), y una recuperación fuerte en el último trimestre que culmina en diciembre (54,687 envíos y 1.73 M USD, temporada navideña). Aquí se pierde el detalle diario y semanal, pero es el más útil para la planificación y la toma de decisiones gerenciales. Los ingresos siguen el mismo patrón que los envíos en los tres niveles.
