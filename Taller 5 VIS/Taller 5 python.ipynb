```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('techstore_participacion_capitulo5.csv')
df
```

```python
# Canal de posición/longitud
plt.figure(figsize=(7, 4))
plt.bar(df['Linea_Producto'], df['Participacion_Pct'], color='steelblue')
plt.ylabel('Participación (%)')
plt.title('Participación por línea de producto – canal: posición/longitud')
plt.show()
```

```python
# Canal de área: círculos con área proporcional al valor, sin ejes de referencia
plt.figure(figsize=(9, 3))
plt.scatter(range(len(df)), [0] * len(df), s=df['Participacion_Pct'] * 100, color='steelblue', alpha=0.8)
for i, nombre in enumerate(df['Linea_Producto']):
    plt.text(i, -0.6, nombre, ha='center')
plt.xlim(-0.7, len(df) - 0.3)
plt.ylim(-1, 1)
plt.axis('off')
plt.title('Participación por línea de producto – canal: área')
plt.show()
```

### Conclusión

Según la **Ley de Stevens**, la magnitud percibida de un estímulo sigue una ley de potencia $S = I^n$, donde el exponente $n$ depende del canal. Para la **longitud**, $n \approx 1$: lo que vemos es casi exactamente proporcional al valor real. Por eso, en el gráfico de barras, alineadas sobre una escala común, se ve de inmediato que Laptops (34 %) es aproximadamente el doble que Accesorios (18 %) y que Accesorios (18 %) supera a Componentes PC (14 %) por unos 4 puntos.

Para el **área**, $n \approx 0.7$: la percepción está **comprimida** y las diferencias se subestiman. Con los círculos, Laptops no parece el doble de Accesorios y la diferencia entre Accesorios y Componentes PC casi desaparece. Además, al no haber ejes ni una línea base común, solo podemos hacer juicios relativos sin referencia (Ley de Weber), lo que reduce aún más la exactitud.

Esto confirma el ranking de efectividad de Munzner y los experimentos de Cleveland y McGill: la **posición alineada/longitud** está entre los canales de magnitud más exactos, mientras que el **área** queda bastante más abajo. Si lo importante es comparar valores cuantitativos, conviene usar barras.

## Actividad 4. Popout y agrupación con seaborn/matplotlib

```python
pts = pd.read_csv('puntos_popout_capitulo5.csv', sep=';')
# El archivo usa 'Si' sin tilde; se aceptan ambas formas
atipicos = pts[pts['Es_Atipico'].isin(['Sí', 'Si'])]
normales = pts[~pts.index.isin(atipicos.index)]
print(len(pts), 'puntos,', len(atipicos), 'atípicos')
atipicos
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5), sharex=True, sharey=True)

# Versión 1: todos los puntos con el mismo color y forma
ax[0].scatter(pts['X'], pts['Y'], color='gray')
ax[0].set_title('Versión 1: mismo color y forma')

# Versión 2: atípicos resaltados con color y forma distintos
ax[1].scatter(normales['X'], normales['Y'], color='gray')
ax[1].scatter(atipicos['X'], atipicos['Y'], color='red', marker='X', s=150, label='Atípico')
ax[1].set_title('Versión 2: atípicos resaltados (popout)')
ax[1].legend()

for a in ax:
    a.set_xlabel('X'); a.set_ylabel('Y')
plt.tight_layout()
plt.show()
```

### Comparación: ¿cuál aprovecha el popout?

En la **versión 1** todos los puntos comparten color y forma, así que para encontrar los dos atípicos hay que recorrer el gráfico punto por punto y fijarse en la posición. Es una búsqueda en serie, y el tiempo crece con el número de puntos.

La **versión 2** aprovecha el **popout** (procesamiento preatencional): las dos "X" rojas se ven al instante entre los 48 puntos grises, sin tener que buscarlas. Según Munzner, el tiempo para detectarlas no depende del número de distractores, porque el ítem distinto se diferencia en un canal de identidad (el **matiz**) y además en la **forma**. Las dos señales apuntan al mismo ítem, así que se refuerzan. El popout deja de funcionar bien cuando hay que buscar una *combinación* de tres o más canales, pero aquí basta con el color para que el atípico resalte.

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=pts, x='X', y='Y', hue='Grupo', s=70)
plt.title('Agrupación perceptual por color (hue = Grupo)')
plt.show()
```

### Agrupación perceptual por color

Al colorear según `Grupo`, el **matiz** (un canal de identidad, el segundo más efectivo para datos categóricos después de la región espacial) hace que los puntos del mismo color se perciban como un conjunto. Los tres grupos se distinguen al instante, incluso donde quedan cerca o se mezclan en el espacio.

En la versión gris solo teníamos la **proximidad espacial** para agrupar, que según el capítulo es una señal más débil que compartir el mismo nivel de un canal de identidad. El color también muestra que los atípicos pertenecen a los Grupos 1 y 3 aunque estén lejos de su nube: el color agrupa aunque la posición los separe. Como el matiz y la posición son canales **separables**, podemos leer al mismo tiempo dónde está cada punto (X, Y) y a qué grupo pertenece (color) sin que una lectura estorbe a la otra.
